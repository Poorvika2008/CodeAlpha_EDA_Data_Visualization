```python
# TASK 2: EXPLORATORY DATA ANALYSIS (EDA)
# Project: Student Performance Data Analysis

import pandas as pd

# Step 1: Create the dataset
data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kavi"],
    "Maths": [85, 90, 65, 95, 75],
    "Science": [80, 85, 70, 90, 80],
    "English": [75, 88, 60, 92, 78]
}

df = pd.DataFrame(data)

# Step 2: Display the dataset
print("STUDENT DATASET")
print(df)

# Step 3: Check dataset information
print("\nDATASET INFORMATION")
df.info()

# Step 4: Statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())

# Step 5: Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Step 6: Calculate average marks
df["Average"] = df[
    ["Maths", "Science", "English"]
].mean(axis=1)

print("\nSTUDENT AVERAGE MARKS")
print(df[["Name", "Average"]])

# Step 7: Find highest scorer
print("\nHIGHEST PERFORMER")
print(df.loc[df["Average"].idxmax()])

# Step 8: Find lowest scorer
print("\nLOWEST PERFORMER")
print(df.loc[df["Average"].idxmin()])

# Step 9: Subject-wise average
print("\nSUBJECT-WISE AVERAGE")
print(df[["Maths", "Science", "English"]].mean())
```

```python
import matplotlib.pyplot as plt

# Bar chart
plt.figure(figsize=(8, 5))
plt.bar(df["Name"], df["Average"])
plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.show()

# Subject comparison
df[["Maths", "Science", "English"]].mean().plot(
    kind="bar",
    title="Average Marks by Subject"
)
plt.ylabel("Marks")
plt.show()

# Pie chart
plt.figure(figsize=(6, 6))
plt.pie(df["Average"], labels=df["Name"], autopct="%1.1f%%")
plt.title("Student Performance")
plt.show()
```
