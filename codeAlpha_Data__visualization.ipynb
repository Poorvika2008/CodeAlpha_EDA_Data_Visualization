```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kavi"],
    "Maths": [85, 90, 65, 95, 75],
    "Science": [80, 85, 70, 90, 80],
    "English": [75, 88, 60, 92, 78]
}

df = pd.DataFrame(data)

print("STUDENT DATA")
print(df)

print("\nDATA SUMMARY")
print(df.describe())

df["Average"] = df[["Maths", "Science", "English"]].mean(axis=1)

print("\nAVERAGE MARKS")
print(df[["Name", "Average"]])

print("\nHIGHEST SCORER")
print(df.loc[df["Average"].idxmax()])

print("\nLOWEST SCORER")
print(df.loc[df["Average"].idxmin()])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(df["Name"], df["Average"])
plt.title("Student Average Marks")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.show()

df[["Maths", "Science", "English"]].mean().plot(
    kind="bar",
    title="Subject-wise Average Marks"
)
plt.ylabel("Average Marks")
plt.show()

plt.figure(figsize=(6, 6))
plt.pie(df["Average"], labels=df["Name"], autopct="%1.1f%%")
plt.title("Student Performance")
plt.show()
```
